## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# -----------------------------------
# Data Wrangling / Data Cleaning
# -----------------------------------

# 1. Create copies of original datasets (good practice)
titles_df = titles.copy()
credits_df = credits.copy()

# -----------------------------------
# 2. Handling Missing Values
# -----------------------------------

# Fill missing age certification with 'Not Rated'
titles_df['age_certification'].fillna('Not Rated', inplace=True)

# Fill missing seasons with 0 (for movies or unknown)
titles_df['seasons'].fillna(0, inplace=True)

# Fill missing IMDb scores with median
titles_df['imdb_score'].fillna(titles_df['imdb_score'].median(), inplace=True)

# Fill missing IMDb votes with 0
titles_df['imdb_votes'].fillna(0, inplace=True)

# Fill missing TMDB scores with median
titles_df['tmdb_score'].fillna(titles_df['tmdb_score'].median(), inplace=True)

# -----------------------------------
# 3. Removing Duplicates
# -----------------------------------

titles_df.drop_duplicates(inplace=True)
credits_df.drop_duplicates(inplace=True)

# -----------------------------------
# 4. Fixing Data Types
# -----------------------------------

# Convert release_year to integer
titles_df['release_year'] = titles_df['release_year'].astype(int)

# Convert seasons to integer
titles_df['seasons'] = titles_df['seasons'].astype(int)

# -----------------------------------
# 5. Handling Columns with Multiple Values
# (genres & production countries are stored as lists in string format)
# -----------------------------------

# Convert string representation of lists into actual lists
import ast

titles_df['genres'] = titles_df['genres'].apply(lambda x: ast.literal_eval(x) if pd.notnull(x) else [])
titles_df['production_countries'] = titles_df['production_countries'].apply(lambda x: ast.literal_eval(x) if pd.notnull(x) else [])

# -----------------------------------
# 6. Basic Consistency Check
# -----------------------------------

print("Cleaned Titles Dataset Shape:", titles_df.shape)
print("Cleaned Credits Dataset Shape:", credits_df.shape)
````
